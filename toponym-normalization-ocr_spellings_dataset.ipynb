```python
import pandas as pd
import json
from src.name_cleanup import value_to_cleaned_str
```

```python
with open("outputs/json/spellings/ocr_places_spellings.json") as fp:
    spellings_json = json.load(fp)
```

```python
threshold = 0.7
spellings_jsonl = []
for actual, spellings in spellings_json.items():
    if actual == "":
        continue

    actual_cleaned = value_to_cleaned_str(actual)
    all_spellings_of = tuple(
        {"spelling": value_to_cleaned_str(spelling[0]), "actual": actual_cleaned}
        for spelling in spellings
        if spelling[1] >= threshold
    )
    spellings_jsonl += all_spellings_of
```

```python
with open("outputs/json/spellings/ocr_places_spellings.jsonl", "w") as fp:
    fp.writelines(tuple(json.dumps(x, ensure_ascii=False) + "\n" for x in spellings_jsonl))
```
