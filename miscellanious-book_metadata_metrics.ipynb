```python
def filter_paths(paths, start, end, sides=["left", "right"]) -> list:
    return list(
        filter(
            lambda x: (
                int((split_stem := x.stem.split("_"))[-3]) in range(start, end)
                and (split_stem[-2] in sides)
            ),
            paths,
        )
    )
```

```python
filter_paths(Path(
    "data/migration-data-csv-release-v1/akaa/"
    "muuttaneet_1876-1886_mko26"
).glob("*.csv"), 0, 13
)
```
